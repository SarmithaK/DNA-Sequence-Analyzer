```python
dna=input("Enter a DNA sequence:").upper()
if not dna or not all(base in "ATGC" for base in dna):
  print("Invalid DNA sequence")
else:
  print("Valid DNA sequence")
  print("sequence length:",len(dna))
  print("G count:",dna.count("G"))
  print("C count:",dna.count("C"))
  print("A count:",dna.count("A"))
  print("T count:",dna.count("T"))
  G_count=dna.count("G")
  C_count=dna.count("C")
  total=len(dna)
  GC_content=(G_count+C_count)/total*100
  print("GC content:",round(GC_content,2),"%")
  A_count=dna.count("A")
  T_count=dna.count("T")
  total=len(dna)
  AT_content=(A_count+T_count)/total*100
  print("AT content:",round(AT_content,2),"%")
  trans_table = str.maketrans("ATGC", "TACG")
  complementary_sequence = dna.translate(trans_table)
  print("COMPLEMENTARY SEQUENCE:",complementary_sequence)
  reverse_complementary_sequence=complementary_sequence[::-1]
  print("Reverse complement:",reverse_complementary_sequence)
  total_ATGC=(GC_content+ AT_content)
  print("total percentage:",total_ATGC)
```
